# Copernicus Marine subset

Validate Copernicus Marine credentials and download a small Mediterranean sea-surface-temperature subset.

Run the cells in order. Credentials are read from local configuration; this notebook contains no saved credentials.

```python
import copernicusmarine
```

## 1. Validate Copernicus Marine credentials


```python
# Run `copernicusmarine login` in a terminal before this cell.
try:
    valid = copernicusmarine.login(check_credentials_valid=True)
except Exception as error:
    print("✗ Credentials could not be checked")
    print(
        "\nCheck your internet connection, then run `copernicusmarine login` "
        "in a terminal and sign in with your Copernicus Marine account. "
        f"Validation detail: {error}"
    )
    raise SystemExit(1) from error

print("✓ Credentials valid" if valid else "✗ Credentials missing or invalid")

if not valid:
    raise RuntimeError(
        "Run `copernicusmarine login` in a terminal, sign in with your "
        "Copernicus Marine account, then run this notebook again."
    )
```

## 2. Define the Mediterranean SST subset


```python
# These parameters can be compared directly with the product page choices.
subset_request = {
    "dataset_id": "SST_MED_SST_L4_NRT_OBSERVATIONS_010_004_c_V2",
    "dataset_version": "202311",
    "variables": ["analysed_sst"],
    "minimum_longitude": 30.68,
    "maximum_longitude": 36.15,
    "minimum_latitude": 32.32,
    "maximum_latitude": 36.72,
    "start_datetime": "2026-09-01T00:00:00",
    "end_datetime": "2026-09-02T00:00:00",
    "coordinates_selection_method": "strict-inside",
    "netcdf_compression_level": 5,
    "disable_progress_bar": False,
}
subset_request
```

## 3. Download the subset


```python
response = copernicusmarine.subset(**subset_request)

# The response records the locally created file(s) and transfer details.
print("\nDownload completed.")
print(response)
```
